```python
# 1. Install deep-translator
!pip install deep-translator

import ipywidgets as widgets
from IPython.display import display
from deep_translator import GoogleTranslator

# 2. Supported languages mapping
languages_dict = GoogleTranslator().get_supported_languages(as_dict=True)
lang_options = [(lang.capitalize(), code) for lang, code in languages_dict.items()]
lang_options.sort()

# 3. Create UI elements
title = widgets.HTML("<h2>🌐 CodeAlpha Language Translation Tool</h2>")
text_input = widgets.Textarea(
    placeholder='Type your text here...',
    description='Input Text:',
    layout=widgets.Layout(width='90%', height='100px')
)

src_lang = widgets.Dropdown(options=[('Auto Detect', 'auto')] + lang_options, value='auto', description='From:')
dest_lang = widgets.Dropdown(options=lang_options, value='es', description='To:')
translate_btn = widgets.Button(description="Translate Text", button_style='primary')

output_text = widgets.Textarea(
    placeholder='Translation will appear here...',
    description='Output:',
    layout=widgets.Layout(width='90%', height='100px'),
    disabled=True
)

# 4. Translation logic
def on_translate_clicked(b):
    if text_input.value.strip():
        translated = GoogleTranslator(source=src_lang.value, target=dest_lang.value).translate(text_input.value)
        output_text.value = translated

translate_btn.on_click(on_translate_clicked)

# 5. Display UI
display(title, src_lang, dest_lang, text_input, translate_btn, output_text)
```
