```python
# Step 1: Install dependencies
!pip install scikit-learn nltk ipywidgets

import ipywidgets as widgets
from IPython.display import display
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Step 2: FAQ Knowledge Base
faq_data = {
    "What is CodeAlpha?": "CodeAlpha is an online platform offering internships and tech education.",
    "How do I submit my internship project?": "Upload your code to GitHub with repo name CodeAlpha_ProjectName and post a LinkedIn video.",
    "When will I get my completion certificate?": "Certificates are issued after reviewing your project repo and LinkedIn video submission.",
    "Can I choose which tasks to complete?": "Yes, you can complete any 2 or 3 tasks from the assigned task list.",
    "What languages are supported for AI tasks?": "Python is standard, using libraries like NLTK, Scikit-learn, or PyTorch."
}

questions = list(faq_data.keys())
answers = list(faq_data.values())

# Step 3: Train simple TF-IDF Model
vectorizer = TfidfVectorizer().fit(questions)
question_vectors = vectorizer.transform(questions)

def get_faq_response(user_query):
    query_vec = vectorizer.transform([user_query])
    similarity_scores = cosine_similarity(query_vec, question_vectors)
    max_idx = similarity_scores.argmax()

    if similarity_scores[0][max_idx] < 0.2:
        return "I'm sorry, I don't have an answer for that specific question in my FAQ database."
    return answers[max_idx]

# Step 4: Simple Chat UI
chat_title = widgets.HTML("<h2>🤖 CodeAlpha FAQ Chatbot</h2>")
query_box = widgets.Text(placeholder="Ask a question (e.g. When will I get my certificate?)...", layout=widgets.Layout(width='80%'))
send_btn = widgets.Button(description="Ask", button_style="success")
chat_output = widgets.Output()

def on_ask(b):
    with chat_output:
        if query_box.value.strip():
            user_text = query_box.value
            print(f"You: {user_text}")
            print(f"Bot: {get_faq_response(user_text)}\n")
            query_box.value = ""

send_btn.on_click(on_ask)
display(chat_title, widgets.HBox([query_box, send_btn]), chat_output)
```
