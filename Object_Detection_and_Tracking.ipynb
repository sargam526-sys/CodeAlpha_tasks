```python
# 1. Install Ultralytics YOLO & OpenCV dependencies
!pip install ultralytics opencv-python-headless

import cv2
from ultralytics import YOLO
from IPython.display import Video, display

# 2. Download a sample traffic/pedestrian video to detect and track
!wget -O sample_video.mp4 "https://github.com/intel-iot-devkit/sample-videos/raw/master/face-demographics-walking.mp4"

# 3. Load a pre-trained lightweight YOLO model (YOLOv8 Nano)
model = YOLO("yolov8n.pt")

# 4. Run Object Detection and Object Tracking (Built-in ByteTrack/SORT tracking)
# persist=True ensures unique track IDs are maintained across video frames
results = model.track(
    source="sample_video.mp4",
    show=False,
    save=True,
    persist=True,
    conf=0.3
)

print("\nTracking complete! Processed video saved in runs/detect/track/")
```
